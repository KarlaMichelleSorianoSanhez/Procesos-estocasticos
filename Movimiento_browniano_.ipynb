# Movimiento Browniano Discretizado

Sea $W(t)$ un movimiento browniano estándar definido en el intervalo $[0,T]$.

Para simular una trayectoria se divide el intervalo temporal en $L$ subintervalos de igual longitud.

$
\delta t=\frac{T}{L}.
$

Definimos los instantes

$
t_i=i\delta t,
\qquad
i=0,1,\ldots,L.
$

Por la propiedad de los incrementos gaussianos del movimiento browniano,

$$
W(t_{i+1})-W(t_i)\sim N(0,\delta t).
$$

Como una variable aleatoria con distribución $N(0,\delta t)$ puede escribirse como

$$
\sqrt{\delta t}\,\xi_i,
\qquad
\xi_i\sim N(0,1),
$$

se obtiene la relación recursiva

$$
W_{i+1}=W_i+\sqrt{\delta t}\,\xi_i.
$$

Además,

$$
W_0=W(0)=0.
$$
Por lo tanto, generando sucesivamente variables normales estándar independientes $xi_i$, es posible construir una aproximación discreta de una trayectoria del movimiento browniano.

# Actividad 1. Implementación del pseudocódigo

Se implementa el pseudocódigo propuesto para simular una trayectoria del movimiento browniano discretizado.

Para ello se fija

$$
T=1,
\qquad
L=100,
$$

por lo que el tamaño de paso es

$$
\delta t=\frac{T}{L}=0.01.
$$

Se toma como condición inicial

$$
W_0=0.
$$

Posteriormente, para cada paso $i=0,\ldots,L-1$, se genera una variable aleatoria

$$
\xi_i\sim N(0,1),
$$

y se actualiza la trayectoria mediante

$$
W_{i+1}=W_i+\sqrt{\delta t}\,\xi_i.
$$

Finalmente se grafica la trayectoria obtenida.

```python
# ==========================================================
# Actividad 1
# Implementación del pseudocódigo de movimiento browniano
# ==========================================================

import numpy as np
import matplotlib.pyplot as plt


# Parámetros
T = 1
L = 100

# Paso temporal
dt = T / L

# Vector de tiempos
t = np.linspace(0, T, L + 1)

# Condición inicial
W = np.zeros(L + 1)

# ----------------------------------------------------------
# Pseudocódigo:
#
# Para i = 0,...,L-1
#   Generar xi_i ~ N(0,1)
#   W(i+1)=W(i)+sqrt(dt)*xi_i
# ----------------------------------------------------------

for i in range(L):

    xi = np.random.normal(0, 1)

    W[i + 1] = W[i] + np.sqrt(dt) * xi

```

```python
# ----------------------------------------------------------
# Gráfica de UNA trayectoria simulada
# ----------------------------------------------------------

plt.figure(figsize=(10,5))

plt.plot(t, W, linewidth=2)

plt.title("Trayectoria simulada del movimiento browniano")
plt.xlabel("t")
plt.ylabel("W(t)")

plt.grid(True)

plt.show()
```

# Actividad 2. Simulación de 100 trayectorias

Utilizando el pseudocódigo implementado en la actividad anterior, se generan 100 trayectorias independientes del movimiento browniano.

En cada trayectoria se utiliza la relación recursiva

$$
W_{i+1}=W_i+\sqrt{\delta t}\,\xi_i,
$$

donde

$$
\xi_i\sim N(0,1).
$$

Cada trayectoria comienza en

$$
W_0=0
$$

y se construye a partir de incrementos normales independientes.

Además, se representan las funciones

$$
f(t)=\sqrt{t}
$$

y

$$
g(t)=-\sqrt{t}.
$$

Estas curvas sirven como referencia para visualizar la dispersión de las trayectorias simuladas conforme aumenta el tiempo.

```python
# ==========================================================
# Actividad 2
# Simulación de 100 trayectorias del movimiento browniano
# ==========================================================

N = 100

# Crear figura
plt.figure(figsize=(10,6))


# Generar las 100 trayectorias

for k in range(N):

    W = np.zeros(L + 1)

    for i in range(L):

        xi = np.random.normal(0,1)

        W[i+1] = W[i] + np.sqrt(dt)*xi

    plt.plot(t, W, alpha=0.25)


# Graficar sqrt(t) y -sqrt(t)
plt.plot(t, np.sqrt(t), linewidth=3,
         label=r'$\sqrt{t}$')

plt.plot(t, -np.sqrt(t), linewidth=3,
         label=r'$-\sqrt{t}$')

plt.title("100 trayectorias del movimiento browniano")
plt.xlabel("t")
plt.ylabel("W(t)")
plt.legend()
plt.grid(True)

plt.show()

```
